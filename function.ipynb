### 2.5 Keyword Arguments


```python
def adder(a,b,c):
    print(f"{a}, {b}, {c}")
    
adder(a=10, c=20, b=44) #keyword 
```

```python
def adder(**kwargs):

    print(kwargs)

adder(a=10, c=20, b=44, n_estimator=50)
```

```python
def adder(**kwargs):
    for key,value in kwargs.items():



        print(f"{key}:{value}")

adder(a=10, c=20, b=44, n_estimator=50)
```

### problem 2
Write a Python generator function called batch_generator. It should take a list of numbers (representing data samples) and a batch_size. The generator should yield one batch at a time as a list.



```python
def batch_generator(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

# Usage
data = [1, 2, 3, 4, 5, 6, 7, 8]
batch_size = 3

for batch in batch_generator(data, batch_size):
    print(batch)
```

```python
def batch_generator(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

# Usage
data = [1, 2, 3, 4, 5, 6, 7, 8]
batch_size = 3

x = batch_generator(data, batch_size) #bikolpo
print(next(x))  
print(next(x))  
print(next(x))  
```

### problem 3
Given a list [1,2,3,4], use map() and lambda to create a new list with squares of each number.

Sample Input:
 No input

Sample Output:
[1, 4, 9, 16]


```python
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, numbers))
print(squares)
```

### problem 4
 Write a function power(base, exponent) that returns the result of base raised to exponent. If the exponent is not given, it should calculate the square.
Sample Input:
5
Sample Output:
25
Sample Input 2:
5 3
Sample Output 2:
125


```python
def power(base, exponent=2):
    return base ** exponent

# Sample 1: only base given → squares by default
print(power(5))

# Sample 2: both base and exponent given
print(power(5, 3))
```

### problem 5
Given a list [1,2,3,4,5,6], use filter() to select even numbers and map() with lambda to square them.
Sample Input:
 No input
Sample Output:
[4, 16, 36]


```python
numbers = [1, 2, 3, 4, 5, 6]
result = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, numbers)))
print(result)
```
