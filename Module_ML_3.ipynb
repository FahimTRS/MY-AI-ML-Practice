```python
#total number of words?
from functools import reduce # reduce er jnno

with open("./sample_data/sample.txt","r") as file:

  strings = file.readlines()

#total line koita?
  total_line = len(strings) #total line koita?
#word ber koro
  number_words = list(map(lambda x: len(x.split()), strings ))
#total word
  total_number_of_words = reduce(lambda x,y : x+y , number_words)

#cleaning process
#1. /new line  delete

  strings = list(map(str.strip, strings))

  #2. space delete

  strings = list(map(lambda x : x.replace(" ",""),strings))

  #character ber korbo

  character_number = list(map(lambda x: len(x), strings ))

  #total number of character

  character_number_total = reduce(lambda x,y : x+y , number_words)


#print(character_number_total)

#store koro new file

with open("./sample_data/counter_of_string.txt","w") as file:

  file.write(f"total line:{total_line}\n total number of words:{total_number_of_words}\n total character:{character_number_total}")
```

### Write a function create_file(filename, content) that creates a new text file with the given filename and writes the content into it. Call it using positional arguments.
📥 Sample Input:
"notes.txt", "Hello, this is my first file."
📤 Sample Output (notes.txt will contain):
Hello, this is my first file.

```python
def create_file(filename, content):
    with open(filename, 'w') as file:
        file.write(content)

create_file("notes.txt", "Hello, this is my first file.")
```

📥 Sample Input (file: data.txt):
Apple

1: Apple
2: Banana
3: Cherry

```python
with open("data.txt", "r") as file:
    text = file.read()
```

```python
import string

# Open and read the file
with open("./sample_data/data.txt", "r") as file:
    text = file.read()

# Convert to lowercase (ignore case sensitivity)
text = text.lower()

# Remove punctuation marks
for char in string.punctuation:
    text = text.replace(char, "")

# Split into words
words = text.split()

# Count frequency of each word
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Find the most frequently occurring word
most_frequent = max(word_count, key=lambda k: word_count[k])

print(most_frequent)
```
