 Problem 2 - C - Even, Odd, Positive and Negative

```python
n = int(input())

inp = input()
numbers = inp.split()

positive = 0; negative = 0; even = 0; odd = 0

for i in range(n):

    x = int(numbers[i])

    #positive or negative
    if x > 0:
            positive +=1
    elif x < 0:
            negative +=1 
    
    #odd or even number
    if x % 2 != 0:
            odd += 1
    else:    #why else ?
            even += 1

print ("even:", even)
print ("odd:", odd)
print ("positive:", positive)
print ("negative:", negative)
```
