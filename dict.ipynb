### 5.4 Dictionary Methods

```python
#printing default values
print(dic) 

print(dic.get("math_marks"))
print(dic.get("math_marks",0))
```

```python
# adding a value

dic ={ "name" : "Phitron" , "age": 20 , "address" : "Dhaka" , "numbers":[10,20,30],"age":30}

print(dic)
dic["math_marks"]= 30 
print(dic) 
dic.update({"english_marks": 40}) 

print(dic) 
```

```python
# deleting 

del dic["math_marks"]
print(dic) 
```

```python
dic_2=dic.copy()
print(dic_2)
```

### Iterating over Dictionary

```python
dic ={ "name" : "Phitron" , "age": 20 , "address" : "Dhaka" , "numbers":[10,20,30],"age":30}

keys = dic.keys()

print(keys)

dic["math_numebers"] = 30 

print(keys) 

values = dic.values() 

print(values) 

dic["math_numebers"] = 40 
print(values) 

items = dic.items() 

print(items)
```

```python
# iteration 
for key,value in dic.items():
    print(key,value)
```

### 5.6 Dictionary Comprehension

```python
square = {x:x**2  for x in range(1,11) if x%2==0}

print(square) 
```

```python
Co_ordinates = [(10,10.5),(20.5,192) , ( 101,102) ]
Locations =["dhaka" , "chattogram" , "sylhet"]

exact_location = {co_or : loc for co_or, loc in zip(Co_ordinates,Locations)}

print(exact_location) #zip(key, value)
```
